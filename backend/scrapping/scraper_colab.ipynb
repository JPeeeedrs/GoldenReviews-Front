```python
import os
import zipfile
from google.colab import files

def fazer_backup_projeto():
    nome_zip = "backup_golden_reviews.zip"

    # Lista com os ficheiros principais E os ficheiros temporários do SQLite
    arquivos_importantes = [
        "cache.db",            # O seu banco de dados principal
        "cache.db-wal",        # NOVO: Os dados mais recentes gravados na memória
        "cache.db-shm",        # NOVO: Índice do banco de dados
        "scraper.log",         # O registo de sucessos e erros
        "games.json",
        "app.py",
        "pipeline_online.py",
        "llm_summary.py",
        "populate_db.py"
    ]

    print(f"📦 A preparar o pacote '{nome_zip}'...")

    # Cria o arquivo ZIP
    with zipfile.ZipFile(nome_zip, 'w', zipfile.ZIP_DEFLATED) as zipf:
        for arquivo in arquivos_importantes:
            if os.path.exists(arquivo):
                zipf.write(arquivo)
                print(f"✅ Adicionado ao ZIP: {arquivo}")
            else:
                # É normal que o -wal e o -shm não existam se a base de dados tiver
                # sido fechada corretamente, por isso apenas ignoramos se não estiverem lá.
                print(f"⚠️ Não encontrado (ignorado): {arquivo}")

    print("\n⬇️ A iniciar o download para a sua máquina...")
    files.download(nome_zip)

# Executa a função
fazer_backup_projeto()
```
